```python
# ============================================================
# 1. Loading the Dataset (FashionMNIST)
# ============================================================
# A Dataset stores the samples and their corresponding labels.
# - root: directory where data is saved
# - train: True for training set (60,000 samples), False for test set (10,000 samples)
# - download: downloads the data if not present in root
# - transform: converts PIL Images to torch.FloatTensor scaled between [0.0, 1.0]

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import datasets
from torchvision.transforms import ToTensor
import matplotlib.pyplot as plt 

# Load training data with ToTensor() so raw images are converted into PyTorch Tensors
training_data = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor()
)

# Load testing data
testing_data = datasets.FashionMNIST(
    root="data",
    train=False, 
    download=True,
    transform=ToTensor()
)

print(f"Training samples: {len(training_data)}")
print(f"Testing samples: {len(testing_data)}")
```

```python
# ============================================================
# 2. Visualizing Dataset Samples with Matplotlib
# ============================================================
# Mapping class indices (0 to 9) to human-readable clothing labels
labels_map = {
    0: "T-Shirt",
    1: "Trouser",
    2: "Pullover",
    3: "Dress",
    4: "Coat",
    5: "Sandal",
    6: "Shirt",
    7: "Sneaker",
    8: "Bag",
    9: "Ankle Boot",
}

figure = plt.figure(figsize=(8, 8))
cols, rows = 3, 3

for i in range(1, cols * rows + 1):
    # Pick a random sample index from training data
    sample_idx = torch.randint(len(training_data), size=(1,)).item()
    img, label = training_data[sample_idx]
    
    figure.add_subplot(rows, cols, i)
    plt.title(labels_map[label])
    plt.axis("off")
    # img is of shape [1, 28, 28] -> squeeze() removes channel dim to shape [28, 28] for imshow
    plt.imshow(img.squeeze(), cmap="gray")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 3. Wrapping Dataset with DataLoader (Batching & Shuffling)
# ============================================================
# While Dataset accesses elements one-by-one, DataLoader:
# 1. Batches data (e.g. batch_size=64 samples per step)
# 2. Shuffles data at each epoch (shuffle=True)
# 3. Loads data in parallel using multiprocessing (num_workers)

train_dataloader = DataLoader(training_data, batch_size=64, shuffle=True)
test_dataloader = DataLoader(testing_data, batch_size=64, shuffle=False)

# Iterate through one mini-batch using next(iter(...))
train_features, train_labels = next(iter(train_dataloader))

print(f"Feature batch shape [Batch, Channels, Height, Width]: {train_features.size()}")
print(f"Labels batch shape [Batch]: {train_labels.size()}")

# Display the first image from this mini-batch
img = train_features[0].squeeze()
label = train_labels[0].item()
plt.imshow(img, cmap="gray")
plt.title(f"First image in batch: {labels_map[label]}")
plt.axis("off")
plt.show()
```
