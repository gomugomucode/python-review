```python
import torch
import numpy as np

# Create a tensor filled with zeros (4 rows and 5 columns)
zero_tensor = torch.zeros(4, 5)  

# Create a tensor filled with ones (4 rows and 5 columns)
one_tensor = torch.ones(4, 5)

# Create a tensor with random numbers uniformly distributed between 0 and 1 (4 rows and 6 columns)
rand_tensor = torch.rand(4, 6)

# Define data as nested lists (2 rows, 3 columns)
data = [[1, 2, 3], [111, 43, 6]]

# Create a 2D tensor from the existing Python data
x = torch.tensor(data)

# Uncomment to inspect the created tensors:
# print(zero_tensor)
# print(one_tensor)
# print(rand_tensor)
print(x)
```

```python
# Inspect the dimensions / shape of the tensor:
# - x.shape: attribute (standard in NumPy and PyTorch)
# - x.size(): PyTorch method
# Both return torch.Size([2, 3]) indicating 2 rows and 3 columns
print(x.size())
print(x.shape)
```

```python
# Common Tensor Attributes

# Create a random tensor of shape (3, 5)
tensor = torch.rand(3, 5)

# Print the tensor values
print("Tensor values:\n", tensor)

# 1. Dimensions / Shape of the tensor:
print("\nSize (method):", tensor.size())   # returns torch.Size([3, 5])
print("Shape (attribute):", tensor.shape)  # returns torch.Size([3, 5])

# 2. Data type of elements (default is torch.float32 for float tensors):
print("Data type:", tensor.dtype)

# 3. Hardware device where tensor is stored ('cpu' or 'cuda'/'mps'):
print("Device:", tensor.device)
```

```python
# Operations on Tensors: Indexing, Slicing & Mutation

# 1. Device check: Check for available hardware accelerator (GPU/MPS/etc.)
if torch.accelerator.is_available():
    device = torch.accelerator.current_accelerator()
    print(f"Using accelerator: {device}")
else:
    device = "cpu"
    print("Using CPU")

# 2. Create a 3x3 tensor filled with 1s
tensor = torch.ones(3, 3, device=device)
print("\nInitial Tensor (3x3 ones):\n", tensor)

# 3. NumPy-like Indexing and Slicing:
# Access the first row (index 0)
print("\nFirst row (tensor[0]):\n", tensor[0])

# Access the last column (index 2 across all rows)
print("\nLast column (tensor[:, 2]):\n", tensor[:, 2])

# 4. In-place Value Modifications:
# Set the entire 3rd row (index 2) to 0
tensor[2] = 0
print("\nAfter setting row 2 to 0 (tensor[2] = 0):\n", tensor)

# Set the 2nd row (slice 1:2) to 33
tensor[1:2] = 33
print("\nAfter setting row slice 1:2 to 33 (tensor[1:2] = 33):\n", tensor)
```

```python
# Joining / Concatenating Tensors

# In PyTorch, use `dim` (not `ndim`) to specify the concatenation dimension:
# - dim=0: concatenate vertically (along rows)
# - dim=1: concatenate horizontally (along columns)

# Concatenating along columns (dim=1)
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print("Concatenated along columns (dim=1):\n", t1)
print("Shape of t1:", t1.shape)
```

```python
# ==========================================
# Arithmetic Operations in PyTorch
# ==========================================

# 1. MATRIX MULTIPLICATION (Linear algebra product)
# Note: `tensor.T` returns the transpose of a tensor.
# y1, y2, y3 will compute the exact same matrix multiplication result:

# Option A: Python '@' operator (most common & concise)
y1 = tensor @ tensor.T

# Option B: Tensor method .matmul()
y2 = tensor.matmul(tensor.T)

# Option C: torch.matmul function with pre-allocated `out=` tensor
y3 = torch.rand_like(y1)
torch.matmul(tensor, tensor.T, out=y3)

print("--- Matrix Multiplication (tensor @ tensor.T) ---")
print("y1:\n", y1)
print("Do y1, y2, y3 have identical values?", torch.equal(y1, y2) and torch.equal(y1, y3))


# 2. ELEMENT-WISE MULTIPLICATION (Hadamard product)
# Multiplies elements at matching positions (a_ij * b_ij).
# z1, z2, z3 will compute the exact same element-wise product:

# Option A: Standard '*' operator
z1 = tensor * tensor

# Option B: Tensor method .mul()
z2 = tensor.mul(tensor)

# Option C: torch.mul function with pre-allocated `out=` tensor
z3 = torch.rand_like(tensor)
torch.mul(tensor, tensor, out=z3)

print("\n--- Element-wise Multiplication (tensor * tensor) ---")
print("z1:\n", z1)
print("Do z1, z2, z3 have identical values?", torch.equal(z1, z2) and torch.equal(z1, z3))
```

```python
# ==========================================
# 1. One-Element Tensors & Aggregations
# ==========================================

# Aggregate function: .sum() computes the sum of all elements, returning a 0-dim scalar tensor
agg = tensor.sum()
print("Tensor representation:", agg)

# .item(): Extracts the single numerical value into a standard Python number (float/int).
# Note: .item() only works on tensors with exactly ONE element.
agg_items = agg.item()
print("Python scalar value:", agg_items, "| Type:", type(agg_items))
```

```python
# ==========================================
# 2. In-Place Operations
# ==========================================

# In PyTorch, operations with a trailing underscore '_' are performed in-place.
# This modifies the tensor directly in memory without allocating a new tensor (saves memory).
# Examples: .add_(), .sub_(), .mul_(), .zero_()

print("Before in-place add:\n", tensor)

# Adds 10 directly to each element in 'tensor'
tensor.add_(10)

print("\nAfter in-place add (tensor.add_(10)):\n", tensor)
```

```python
# ==========================================
# 3. Converting PyTorch Tensor to NumPy Array
# ==========================================

# Create a 1D tensor of five 1s on CPU
t = torch.ones(5)
print("PyTorch Tensor (t):\n", t)

# .numpy() converts a CPU tensor to a NumPy ndarray.
# IMPORTANT: The tensor and NumPy array share the exact same underlying memory (zero-copy).
n = t.numpy()
print("\nNumPy Array (n):\n", n)
```

```python
# ==========================================
# 4. Modifying Tensor Reflects in NumPy Array
# ==========================================

# Because 't' and 'n' share memory on CPU, modifying 't' in-place automatically modifies 'n' as well.
t.add_(10)

print("Updated Tensor (t):\n", t)
print("\nUpdated NumPy Array (n):\n", n)
```

```python
# ==========================================
# 5. Converting NumPy Array to PyTorch Tensor
# ==========================================

# Create a NumPy array with three 1s (default dtype is float64)
n = np.ones(3)

# torch.from_numpy() creates a tensor directly referencing the array's memory (zero-copy)
t = torch.from_numpy(n)

print("NumPy Array (n):\n", n)
print("\nPyTorch Tensor (t):\n", t)
```

```python
# ==========================================
# 6. Modifying NumPy Array Reflects in Tensor
# ==========================================

# np.add(n, 1, out=n) performs an in-place addition on 'n' (storing the result back in 'n').
# Note:
# - Both operands 'n' and '1' are required: np.add(operand1, operand2, out=destination)
# - You can also write: n += 1
# - AVOID 'n = n + 1', which creates a brand new array and breaks the shared memory connection with 't'!
np.add(n, 1, out=n)

print("Updated NumPy Array (n):\n", n)
print("\nUpdated PyTorch Tensor (t):\n", t)
```
